```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'staff':['A','B','C','D','E','F'],
                    'issues':[18,22,29,25,14,11],
                    'assists':[4,5,5,4,8,12],
                    'rebounds':[10,6,4,6,3,5]})
```

```python
print(df1)
```
