**Ranking**

```python
import pandas as pd
import numpy as np
```

```python
obj = pd.Series([7,-5,7,4,2,0,4])
```

```python
obj.rank()
```

```python
obj.rank(method = 'first')
```

```python
obj.rank(ascending = False,method = 'max')
```

```python
frame = pd.DataFrame({'b':[4.3,7,-3,2],'a':[0,1,0,1],'c':[-2,5,8,-2.5]})
```

```python
frame.rank(axis = 'columns')
```

```python
df = pd.DataFrame([[1.4,np.nan],
                   [7.1,-4.5],
                   [np.nan,np.nan],
                   [0.75,-1.3]],
                  index = ['a','b','c','d'],
                  columns = ['one','two'])
                  
```

```python
df.sum()
```

```python
df.sum(axis=1)
```

```python
df.mean(axis=1,skipna=False)
```

```python
df.idxmax()
```

```python
df.cumsum(axis=1)
```

```python
df.cumsum()
```

```python
df.describe()
```
